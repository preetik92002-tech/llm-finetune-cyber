# Smoke test on Colab

Runs `src/train.py` with `configs/smoke_test.yaml` on a few examples to check the pipeline.
Set a GPU runtime first: Runtime > Change runtime type > T4 GPU.

Add these in the Colab Secrets panel (key icon) and allow notebook access:
- `HF_TOKEN` (Hugging Face token)
- `WANDB_API_KEY` (optional; without it, Weights and Biases logging stays off)

```
# Get the code.
!git clone https://github.com/preetik92002-tech/llm-finetune-cyber.git
%cd llm-finetune-cyber
```

```
# Install the packages.
!pip install -q -r requirements.txt
```

```
import os
from google.colab import userdata

# Read tokens from Colab Secrets. Never paste a token into the notebook.
os.environ["HF_TOKEN"] = userdata.get("HF_TOKEN")
try:
    os.environ["WANDB_API_KEY"] = userdata.get("WANDB_API_KEY")
except Exception:
    print("No WANDB_API_KEY secret: W and B logging stays off.")
```

```
# Run the smoke test.
!python src/train.py --config configs/smoke_test.yaml
```
